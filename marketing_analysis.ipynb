## 🏆 Section 5 — Platform Analysis

### Platform Performance Table

````python
platform_summary = (
    df.groupby("Platform", as_index=False)
      .agg(
          Clicks=("Clicks","sum"),
          Conversions=("Conversions","sum"),
          Cost=("Cost","sum"),
          Conversion_Rate=("Conversion_Rate","mean"),
          CPC=("CPC","mean"),
          CPA=("CPA","mean"),
          Conversion_Efficiency=("Conversion_Efficiency","mean"),
      )
)
fig_conv = px.bar(
    platform_summary.sort_values("Conversions", ascending=False),
    x="Platform",
    y="Conversions",
    title="Conversions by Platform",
    text="Conversions",
)
fig_conv.update_layout(yaxis_title="Conversions", xaxis_title="Platform", uniformtext_minsize=8, uniformtext_mode="hide")
fig_cost = px.bar(
    platform_summary.sort_values("Cost", ascending=False),
    x="Platform",
    y="Cost",
    title="Cost by Platform",
    text="Cost",
)
fig_cost.update_layout(yaxis_title="Cost", xaxis_title="Platform")
fig_cr = px.bar(
    platform_summary.sort_values("Conversion_Rate", ascending=False),
    x="Platform",
    y="Conversion_Rate",
    title="Average Conversion Rate by Platform",
    text="Conversion_Rate",
)
fig_cr.update_layout(yaxis_title="Conversion Rate", xaxis_title="Platform")
fig_cpa_cpc = make_subplots(rows=1, cols=2, subplot_titles=("Average CPA by Platform", "Average CPC by Platform"))
fig_cpa_cpc.add_trace(go.Bar(x=platform_summary["Platform"], y=platform_summary["CPA"], name="CPA"), row=1, col=1)
fig_cpa_cpc.add_trace(go.Bar(x=platform_summary["Platform"], y=platform_summary["CPC"], name="CPC"), row=1, col=2)
fig_cpa_cpc.update_layout(title_text="CPA and CPC by Platform", showlegend=False)
fig_donut = px.pie(platform_summary, values="Conversions", names="Platform", title="Conversion Share by Platform")
fig_score = px.bar(
    platform_summary.sort_values("Conversion_Efficiency", ascending=False),
    x="Platform",
    y="Conversion_Efficiency",
    title="Conversion Efficiency by Platform",
    text="Conversion_Efficiency",
)
fig_score.update_layout(yaxis_title="Conversion Efficiency", xaxis_title="Platform")
monthly_summary = (
    df.groupby("Start_YearMonth", as_index=False)
      .agg(
          Conversions=("Conversions","sum"),
          Cost=("Cost","sum"),
          Clicks=("Clicks","sum"),
      )
      .sort_values("Start_YearMonth")
)
fig_mconv = px.line(monthly_summary, x="Start_YearMonth", y="Conversions", title="Monthly Conversions", markers=True)
fig_mcost = px.line(monthly_summary, x="Start_YearMonth", y="Cost", title="Monthly Cost", markers=True)
fig_mact = px.line(monthly_summary, x="Start_YearMonth", y="Clicks", title="Monthly Click Activity", markers=True)
fig_top10 = px.bar(
    df.nlargest(10, "Conversions"),
    x="Campaign_Name",
    y="Conversions",
    title="Top 10 Campaigns by Conversions",
    text="Conversions",
)
fig_top10.update_layout(xaxis_tickangle=-45)
fig_top10cr = px.bar(
    df.nlargest(10, "Conversion_Rate"),
    x="Campaign_Name",
    y="Conversion_Rate",
    title="Top 10 Campaigns by Conversion Rate",
    text="Conversion_Rate",
)
fig_top10cr.update_layout(xaxis_tickangle=-45)
duration_summary = df["Duration_Group"].value_counts().reset_index(name="Count").rename(columns={"index":"Duration_Group"})
fig_dur = px.bar(duration_summary, x="Duration_Group", y="Count", title="Campaigns by Duration Group", text="Count")
fig_corr = px.imshow(
    df.select_dtypes(include=[float, int]).corr(),
    text_auto=True,
    title="Correlation Matrix for Numeric Variables",
    labels={"x":"Metric", "y":"Metric", "color":"Correlation"},
)
````

**Composite Score Formula:**

```
Score = 0.40 × norm(Conversion Rate)
      + 0.30 × norm(Total Conversions)
      + 0.20 × norm(Conversion Efficiency)
      − 0.10 × norm(CPA)
```
All components are min-max normalised to [0, 1] before weighting.

## 💡 Section 10 — Key Business Insights & Recommendations

### Top Findings

| # | Insight |
|---|---------|
| 1 | **Google** is the top-performing platform (score 0.862) — highest conversions (498K) and lowest CPA ($24.19) |
| 2 | **Facebook** has the highest conversion rate (10.47%) but ranks #2 due to slightly higher CPA |
| 3 | **Twitter** is the weakest performer (score 0.006) — lowest CR (9.21%) with $4.77M spend |
| 4 | Campaign activity peaked in **Feb 2024** (174 campaigns) and declined 34% by Oct 2024 (115) |
| 5 | **Longer campaigns** (> 12 months) average 40% more conversions than sub-30-day bursts |

---

### Strategic Recommendations

1. **Reallocate Twitter budget (25–30%) to Google** — highest ROI improvement with zero incremental spend
2. **Anchor strategy on Google + Facebook** — protect combined 50–55% budget share
3. **Invest in longer campaign durations** — evergreen campaigns consistently outperform short flights
4. **Plan H2 campaigns proactively** — the 34% Q3/Q4 volume drop is avoidable with advance planning
5. **Scale LinkedIn and Instagram** — solid ~10% CR with room to grow from smaller bases
6. **Implement early-warning rules** — pause campaigns with >500 clicks and <5 conversions after 14 days
